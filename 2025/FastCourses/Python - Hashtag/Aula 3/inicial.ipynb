### Case: Score de Crédito dos Clientes

Você foi contratado por um banco para conseguir definir o score de crédito dos clientes. Você precisa analisar todos os clientes do banco e, com base nessa análise, criar um modelo que consiga ler as informações do cliente e dizer automaticamente o score de crédito dele: Ruim, Ok, Bom

Arquivos da aula: https://drive.google.com/drive/folders/1FbDqVq4XLvU85VBlVIMJ73p9oOu6u2-J?usp=drive_link

```python
import pandas as pd

df = pd.read_csv('clientes.csv');

display(df);

#pt1
```

```python
display(df.info());
#pt2
```

```python
from sklearn.preprocessing import LabelEncoder

codificador1 = LabelEncoder();
df["profissao"] = codificador1.fit_transform(df["profissao"]);

codificador2 = LabelEncoder();
df["mix_credito"] = codificador2.fit_transform(df["mix_credito"]);

codificador3 = LabelEncoder();
df["comportamento_pagamento"] = codificador3.fit_transform(df["comportamento_pagamento"]);
display(df.info());
#pt3
```

```python
y = df["score_credito"];
x = df.drop(columns=["score_credito", "id_cliente"]);

from sklearn.model_selection import train_test_split
x_treino, x_teste, y_treino, y_teste = train_test_split(x, y, test_size=0.2);

#pt4
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

modelo_arvore = RandomForestClassifier();
modelo_kn = KNeighborsClassifier();

modelo_arvore.fit(x_treino, y_treino);
modelo_kn.fit(x_treino, y_treino);
```

```python
previsao_arvore = modelo_arvore.predict(x_teste);
previsao_kn = modelo_kn.predict(x_teste);

from sklearn.metrics import accuracy_score

display(accuracy_score(y_teste, previsao_arvore));
display(accuracy_score(y_teste, previsao_kn));
```

para melhorar a accuracy podemos testar outros modelos, aumentar a base de dados, mudar parametros
